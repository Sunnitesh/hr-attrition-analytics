```python
df["Attrition"]=df["Attrition"].map({"Yes":1,"No":0})
df["OverTime"]=df["OverTime"].map({"Yes":1,"No":0})
df["Gender"]=df["Gender"].map({"Male":1,"Female":0})
```

```python
df = pd.get_dummies(df, columns=['Department', 'JobRole', 'MaritalStatus', 'EducationField', 'BusinessTravel'], drop_first=True)
df.shape
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split

x=df.drop("Attrition",axis=1)
y=df["Attrition"]

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
```

```python
print(y_train.value_counts())
print(y_test.value_counts())
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)
model.fit(x_train, y_train)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
model = LogisticRegression(max_iter=1000)
model.fit(x_train_scaled, y_train)
y_pred = model.predict(x_test_scaled)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1 Score:", f1_score(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

```python
model_balanced=LogisticRegression(max_iter=100,class_weight="balanced")
model_balanced.fit(x_train_scaled , y_train)
y_pred_balanced = model_balanced.predict(x_test_scaled)

print("Classification Report :\n",classification_report(y_test,y_pred_balanced))
print("Confusion Matrix :\n",confusion_matrix(y_test,y_pred_balanced))
```

```python
from sklearn.tree import DecisionTreeClassifier

model_dt=DecisionTreeClassifier(random_state=42,class_weight="balanced")
model_dt.fit(x_train_scaled,y_train)
y_pred_dt=model_dt.predict(x_test_scaled)
print("Classification Report:\n", classification_report(y_test, y_pred_dt))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred_dt))
```

```python
from sklearn.ensemble import RandomForestClassifier

model_rf = RandomForestClassifier(n_estimators=100, random_state=42, class_weight='balanced')
model_rf.fit(x_train_scaled, y_train)
y_pred_rf = model_rf.predict(x_test_scaled)

print("Classification Report:\n", classification_report(y_test, y_pred_rf))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred_rf))
```

```python
# Get probabilities instead of hard predictions
y_pred_rf_proba = model_rf.predict_proba(x_test_scaled)[:, 1]

# Lower threshold from 0.5 to 0.3
y_pred_rf_tuned = (y_pred_rf_proba >= 0.3).astype(int)

print("Classification Report (threshold=0.3):\n", classification_report(y_test, y_pred_rf_tuned))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred_rf_tuned))
```

```python
y_pred_rf_tuned2 = (y_pred_rf_proba >= 0.15).astype(int)

print("Classification Report (threshold=0.15):\n", classification_report(y_test, y_pred_rf_tuned2))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred_rf_tuned2))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

feature_importance = pd.Series(model_rf.feature_importances_, index=x_train.columns)
top_features = feature_importance.sort_values(ascending=False).head(15)

plt.figure(figsize=(10, 6))
top_features.plot(kind='barh')
plt.title('Top 15 Feature Importances - Random Forest')
plt.xlabel('Importance Score')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```
